## Create Cursor

```python
# create cursor
cur = conn.cursor()
cur.execute("SELECT * FROM site")

# A pointer 
```

```python
# You can apply the fetchall()..
cur.df()
```

The cursors dissapears after use! 
Create several in a loop

```python
cur.execute("SELECT * FROM site")
cur.fetchone()[0]
```

```python
# We want to iterate through the rows and pring the site code
cur.execute("SELECT * FROM site")

for row in cur.fetchall():
    print(row[0])
```

```python
cur.execute("SELECT * FROM site")

while True:
    row = cur.fetchone()
    if row == None:
            break
    print(f"Site code: {row[0]}")
```

```python
# You can create table 
cur.execute("""CREATE TEMP TABLE temp_nests 
            AS SELECT * FROM Bird_nests""")
cur.execute("SELECT * FROM temp_nests").fetchall()
```
