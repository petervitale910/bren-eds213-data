```python
conn = duckdb.connect("database/database.duckdb")
```

```python
# query to be executed, filled by user input with python string interpolation
query = "SELECT * FROM Species WHERE Code = '%s'"

# craft sql injection
inject_input = "wolv'; INSERT INTO Personnel (Abbreviation, Name) VALUES ('tswift', 'Taylor Swift');--"

# confirm output
print(query % (inject_input))
```
